```python
pd.crosstab(df["gender"], df["trip_rating"]).plot(kind="bar")
plt.title("Trip Rating by Gender")
plt.xlabel("Gender")
plt.ylabel("Trip Rating")

plt.show()
```

```python
average_fare = df.groupby("service_type")["fare"].mean()
average_fare
```

```python
average_fare.plot(kind="bar")
plt.title("Average Fare by Service Type")
plt.xlabel("Service Type")
plt.ylabel("Average Fare (PKR)")

plt.show()
```
